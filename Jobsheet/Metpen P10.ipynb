#  A. Soal Pengujian Hipotesis Univariat

```python
import numpy as np
from scipy import stats

# Data hasil percobaan algoritma baru
data = np.array([118, 117, 120, 119, 116])

# Parameter
mu_quicksort = 120   # Rata-rata QuickSort
alpha = 0.05         # Tingkat signifikansi

# Hitung rata-rata dan simpangan baku sampel
mean_sample = np.mean(data)
std_sample = np.std(data, ddof=1)  # ddof=1 untuk simpangan baku sampel
n = len(data)

# Hitung nilai t-statistic
t_statistic = (mean_sample - mu_quicksort) / (std_sample / np.sqrt(n))

# Hitung p-value (uji satu arah: apakah lebih kecil dari 120)
p_value = stats.t.cdf(t_statistic, df=n-1)

# Cetak hasil
print("=== UJI T SATU SAMPEL ===")
print(f"Data sampel: {data}")
print(f"Rata-rata sampel = {mean_sample:.2f}")
print(f"Simpangan baku sampel = {std_sample:.4f}")
print("===========================")
print(f"t hitung = {t_statistic:.4f}")
print(f"p-value = {p_value:.4f}")

# Keputusan
if p_value < alpha:
    print(f"Keputusan: Tolak H0 (algoritma baru lebih cepat dari QuickSort)")
else:
    print(f"Keputusan: Gagal menolak H0 (tidak cukup bukti bahwa algoritma baru lebih cepat)")
```

#  B. Soal Pengujian Hipotesis Bivariat

```python
import numpy as np
from scipy import stats

# Data eksperimen
x = np.array([200, 400, 600, 800, 1000])   # Jumlah data training
y = np.array([80, 83, 86, 88, 90])         # Akurasi model (%)

# Hitung korelasi Pearson
r, p_value = stats.pearsonr(x, y)

print("=== UJI KORELASI PEARSON ===")
print(f"Koefisien korelasi (r) = {r:.4f}")
print(f"Nilai p-value (two-tailed) = {p_value:.4f}")

# Keputusan
alpha = 0.05
if p_value < alpha:
    print("Keputusan: Tolak H0 (terdapat hubungan signifikan antara jumlah data dan akurasi)")
else:
    print("Keputusan: Gagal menolak H0 (tidak ada bukti hubungan signifikan)")
```

#  C. Soal Pengujian Hipotesis Multivariat

```python
import pandas as pd
import statsmodels.api as sm

# Data dari soal
data = {
    "LearningRate": [0.001, 0.001, 0.005, 0.005, 0.01],
    "Epoch": [20, 40, 20, 40, 40],
    "BatchSize": [16, 32, 16, 32, 32],
    "Akurasi": [86, 88, 85, 90, 87]
}

# Membuat DataFrame
df = pd.DataFrame(data)

# Variabel independen (X) dan dependen (Y)
X = df[["LearningRate", "Epoch", "BatchSize"]]
Y = df["Akurasi"]

# Menambahkan konstanta (β₀)
X = sm.add_constant(X)

# Membuat model regresi linear berganda
model = sm.OLS(Y, X).fit()

# Menampilkan hasil
print(model.summary())
```

```python
# ==========================================
# Analisis Regresi Linear Berganda
# Kasus: Pengaruh Learning Rate, Epoch, dan Batch Size
# terhadap Akurasi Model YOLOv8
# ==========================================

# Import library yang dibutuhkan
import pandas as pd
import statsmodels.api as sm

# ------------------------------------------
# 1. Membuat dataset sesuai dengan tabel di PDF
# ------------------------------------------
data = {
    'LearningRate': [0.001, 0.001, 0.005, 0.005, 0.01],
    'Epoch': [20, 40, 20, 40, 40],
    'BatchSize': [16, 32, 16, 32, 32],
    'Akurasi': [86, 88, 85, 90, 87]
}

# Membuat DataFrame dari data di atas
df = pd.DataFrame(data)

# ------------------------------------------
# 2. Menentukan variabel independen (X) dan dependen (Y)
# ------------------------------------------
X = df[['LearningRate', 'Epoch', 'BatchSize']]  # variabel bebas
Y = df['Akurasi']                               # variabel terikat

# Menambahkan konstanta (intercept β₀)
X = sm.add_constant(X)

# ------------------------------------------
# 3. Membuat dan menjalankan model regresi linear berganda
# ------------------------------------------
model = sm.OLS(Y, X).fit()

# ------------------------------------------
# 4. Menampilkan hasil ringkasan analisis statistik
# ------------------------------------------
print(model.summary())

# ------------------------------------------
# 5. (Opsional) Menampilkan koefisien regresi saja
# ------------------------------------------
print("\nKoefisien Regresi:")
print(model.params)

print("\nP-Value masing-masing variabel:")
print(model.pvalues)

print("\nKoefisien Determinasi (R²):", model.rsquared)
```
